```python
class Estudante: 
    def __init__(self, nome, faculdade, curso, semestre):
            self.nome = nome 
            self.faculdade = faculdade
            self.curso = curso 
            self.semestre = semestre

    def apresentar(self):
        return f"Ola meu nome é {self.nome}! Estudo {self.curso}, na faculdade {self.faculdade}, ela dura {self.semestre}."
    
a1 = Estudante("Victor", "Fiap", "Engenharia de software", "8 semestres")
a2 = Estudante("Juan", "Fiap", "Engenharia de software", "8 semestres")
a3 = Estudante("Laura", "Fiap", "Engenharia de software", "8 semestres")
a4 = Estudante("Rebeca", "Brascubas", "Engenharia Meca", "10 semestres")

            
print(a1.apresentar())
print(a2.apresentar())
print(a3.apresentar())
print(a4.apresentar())
```

```python
class Retangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2* (self.base + self.altura)

r1 = Retangulo(float(input("Base: ")), float(input("Altura:")))
print (f"A area do r1: {r1.area()} - perimetro {r1.perimetro()}")
```

```python
class IMC:
    def __init__(self, nome, peso, altura):
        self.peso = peso
        self.altura = altura
        self.nome = nome

    def Calculo_IMC(self):
        return self.peso / (self.altura ** 2)

p1 = IMC("Gustavinho", 65, 1.65)
p2 = IMC("Paulo", 70, 1.70)
p3 = IMC("victor", 75, 1.86)

print(f"O {p1.nome} tem {p1.peso}kg e mede {p1.altura}m — IMC: {p1.Calculo_IMC():.2f}")
print(f"O {p2.nome} tem {p2.peso}kg e mede {p2.altura}m — IMC: {p2.Calculo_IMC():.2f}")
print(f"O {p3.nome} tem {p3.peso}kg e mede {p3.altura}m — IMC: {p3.Calculo_IMC():.2f}")
```

```python
class Juros: 
    def __init__(self, nome, capital, taxa, tempo):
        self.nome = nome
        self.capital = capital
        self.taxa = taxa
        self.tempo = tempo

    def montante(self):
        return self.capital * (1 + self.taxa/100) ** self.tempo

    def lucro(self):
        return self.montante() - self.capital


juros = []

for i in range(3):
    valores = Juros(
        input("Nome:"),
        float(input("Capital: ")),
        float(input("taxa: ")),
        int(input("tempo: ")),
    )
    juros.append(valores)

for valor in juros: 
    print(f"A pessoa {valor.nome} deverá receber R${valor.lucro():.2f} de lucro sobre seu montante, R${valor.montante():.2f}")
```

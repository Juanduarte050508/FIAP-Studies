```python
import pandas as pd

dados ={
    "produto" : ["notebook", "mouse", "teclado", "monitor"],
    "vendas" : [1200, 300, 450, 800],
    "lucro" : [300, 50, 80, 200],
    "ano" : [2022, 2022, 2022, "dois mil e vinte dois",]
}
print(dados)

dataframe = pd.DataFrame(dados)

for coluna in ["vendas", "lucro"]:
    try:
        media = dataframe[coluna].mean()
        print(f"Mediada coluna{coluna} é {media}")
    except KeyError:
        print(f"A coluna {coluna} não existe")
```

```python
#2 Crie uma estrutura ocm try/except a qual calcule e traga um erro para calcular a area de um retangulo e circulo com estrutura de input para o usuáriodeve escolher a figura geométrica 

try:
    choice = (input("Escolha se quer a forma geometrica"))
    if choice.lower() == "triangulo":
        try:
            base= float(input("Digite a base do triangulo"))
            altura= float(input("Digite a altura do triangulo"))
            areaTriangulo= (base * altura)/2
            print(areaTriangulo)
        except ValueError:
            print("O valor digitado não é numerico")
    elif choice.lower() == "retangulo":
        try:
            base= float(input("Digite a base do retangulo"))
            altura= float(input("Digite a altura do retangulo"))
            areaRetangulo= (base * altura)
            print(areaRetangulo)
        except ValueError:
            print("O valor digitado não é numerico")
    elif choice.lower() == "circulo":
        try:
            raio= float(input("Digite o Raio do circulo"))
            pi = 3.14
            areaCircular= pi * (raio * raio)
            print(areaCircular)
        except ValueError:
            print("O valor digitado não é numerico")
except:
    print ("Esse forma geometrica não esta inclusa")
```

```python
#EXC 1

class Algoritimo:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario 

    def taxa (self):
        return self.salario * 0.275

    def desconto(self):
     return self.salario -(self.taxa())

Victor = Algoritimo("Victor", 1.60)  
Rafael = Algoritimo("Rafael", 2600)  
Juan = Algoritimo("Juan", 3000)

print(Rafael.taxa(), (Victor.desconto()))
print(Juan.taxa(), (Juan.desconto()))
print(Victor.desconto(), (Rafael.desconto()))
```

```python
#EXC 2

class Algoritimo:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario 

    def taxa (self):
        return self.salario * 0.275

    def desconto(self):
     return self.salario -(self.taxa())

Victor = Algoritimo("Victor", 100)  
Rafael = Algoritimo("Rafael", 2600)  
Juan = Algoritimo("Juan", 3000)
Laura = Algoritimo("Laura", 2100)
Rebeca = Algoritimo("Rebeca", 7000)

#Adição de lista para printar 

lista = [Victor, Rafael, Juan, Laura, Rebeca]

for pessoa in lista:
    print(f"Nome: {pessoa.nome} | Salário: {pessoa.salario} | Taxa: {pessoa.taxa():.2f} | Desconto: {pessoa.desconto():.2f}")
```

```python
# Estrutura com recursão
def it(n):
    if n == 0:
        return 1
    return n * it ( n - 1 )

it()
```

```python
# Estrutura com repetição

def it(n):

    result = 1
    for i in range(1, n+1):
        result *= i

    return result
%time it(500)
```
